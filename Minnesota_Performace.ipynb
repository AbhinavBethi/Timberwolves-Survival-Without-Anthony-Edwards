**Classify every game: Edwards played vs. missed**

```python


EDWARDS_ID = 4594268

# Find Edwards' records in Minnesota's regular-season dataset
edwards = wolves[
    wolves["athlete_id"] == EDWARDS_ID
].copy()

print("Anthony Edwards records:", len(edwards))
print("Unique games with an Edwards record:", edwards["game_id"].nunique())

# Check whether each record says he played
display(
    edwards[
        [
            "game_id",
            "game_date",
            "athlete_display_name",
            "minutes",
            "points",
            "did_not_play",
            "reason"
        ]
    ].sort_values("game_date").head(15)
)
```

```python


# One record per Minnesota game
team_games = (
    wolves.groupby("game_id", as_index=False)
    .agg(
        game_date=("game_date", "first"),
        team_score=("team_score", "first"),
        opponent_score=("opponent_team_score", "first"),
        team_winner=("team_winner", "first")
    )
)

# Add Edwards' status to each game
edwards_status = edwards[
    ["game_id", "minutes", "did_not_play", "reason"]
].copy()

edwards_status = edwards_status.rename(
    columns={
        "minutes": "edwards_minutes",
        "did_not_play": "edwards_did_not_play",
        "reason": "edwards_reason"
    }
)

team_games = team_games.merge(
    edwards_status,
    on="game_id",
    how="left",
    validate="one_to_one"
)


def classify_edwards(row):
    if pd.isna(row["edwards_did_not_play"]):
        return "UNKNOWN"

    if bool(row["edwards_did_not_play"]):
        return "MISSED"

    if pd.notna(row["edwards_minutes"]) and row["edwards_minutes"] > 0:
        return "PLAYED"

    return "UNKNOWN"

team_games["EDWARDS_STATUS"] = team_games.apply(
    classify_edwards,
    axis=1
)

print("Game classification:")
display(
    team_games["EDWARDS_STATUS"]
    .value_counts(dropna=False)
    .rename_axis("status")
    .reset_index(name="games")
)

print("\nGames requiring investigation:")
display(
    team_games[
        team_games["EDWARDS_STATUS"] == "UNKNOWN"
    ][
        [
            "game_date",
            "game_id",
            "edwards_minutes",
            "edwards_did_not_play",
            "edwards_reason"
        ]
    ]
)

print("\nEdwards' missed games:")
display(
    team_games[
        team_games["EDWARDS_STATUS"] == "MISSED"
    ][
        ["game_date", "team_score", "opponent_score",
         "team_winner", "edwards_reason"]
    ].sort_values("game_date")
)
```

**Compare Minnesota's team performance**

```python


valid_games = team_games[
    team_games["EDWARDS_STATUS"].isin(["PLAYED", "MISSED"])
].copy()

valid_games["win"] = valid_games["team_winner"].astype("boolean")

team_summary = (
    valid_games.groupby("EDWARDS_STATUS")
    .agg(
        games=("game_id", "nunique"),
        wins=("win", "sum"),
        avg_points_scored=("team_score", "mean"),
        avg_points_allowed=("opponent_score", "mean")
    )
    .reset_index()
)

team_summary["losses"] = (
    team_summary["games"] - team_summary["wins"]
)

team_summary["win_pct"] = (
    team_summary["wins"] / team_summary["games"] * 100
)

team_summary["avg_point_differential"] = (
    team_summary["avg_points_scored"]
    - team_summary["avg_points_allowed"]
)

team_summary = team_summary[
    [
        "EDWARDS_STATUS",
        "games",
        "wins",
        "losses",
        "win_pct",
        "avg_points_scored",
        "avg_points_allowed",
        "avg_point_differential"
    ]
]

display(
    team_summary.round(2)
)
```

```python


# 1. Count player records and distinct players in each Minnesota game
game_coverage = (
    wolves.groupby(["game_id", "game_date"], as_index=False)
    .agg(
        player_records=("athlete_id", "size"),
        unique_players=("athlete_id", "nunique"),
        players_with_minutes=("minutes", lambda x: x.notna().sum()),
        inactive_records=("did_not_play", lambda x: x.eq(True).sum())
    )
)

unknown_coverage = unknown_games[
    ["game_id", "game_date"]
].merge(
    game_coverage,
    on=["game_id", "game_date"],
    how="left"
).sort_values("game_date")

print("Coverage of the 21 games:")
display(unknown_coverage)

# 2. Check whether Edwards has any records in the regular season
edwards_regular = player_data[
    (player_data["season"] == 2026) &
    (player_data["season_type"] == REGULAR_SEASON_TYPE) &
    (player_data["athlete_id"] == EDWARDS_ID)
].copy()

print("\nEdwards regular-season records:", len(edwards_regular))
print("Unique games with an Edwards record:", edwards_regular["game_id"].nunique())
print("Games with positive minutes:", (edwards_regular["minutes"] > 0).sum())

# 3. Search by name as well as athlete ID
name_matches = player_data[
    (player_data["season"] == 2026) &
    (player_data["season_type"] == REGULAR_SEASON_TYPE) &
    player_data["athlete_display_name"].astype(str).str.contains(
        "Anthony Edwards", case=False, na=False
    )
]

print("\nRecords matching the player name:")
display(
    name_matches[
        [
            "game_id", "game_date", "athlete_id",
            "athlete_display_name", "team_abbreviation",
            "minutes", "points", "did_not_play"
        ]
    ].sort_values("game_date")
)

# 4. Inspect the player records for the first five unknown games
sample_ids = unknown_games.sort_values("game_date")["game_id"].head(5)

print("\nPlayer records for the first five unknown games:")
display(
    wolves[wolves["game_id"].isin(sample_ids)][
        [
            "game_id", "game_date", "athlete_display_name",
            "minutes", "points", "did_not_play"
        ]
    ].sort_values(["game_date", "athlete_display_name"])
)
```

```python

# Minnesota regular-season games
min_games = (
    player_data[
        (player_data["season"] == 2026) &
        (player_data["season_type"] == REGULAR_SEASON_TYPE) &
        (player_data["team_abbreviation"] == "MIN")
    ][["game_id", "game_date"]]
    .drop_duplicates("game_id")
    .copy()
)

# Anthony Edwards' regular-season records
ant_records = player_data[
    (player_data["season"] == 2026) &
    (player_data["season_type"] == REGULAR_SEASON_TYPE) &
    (player_data["athlete_id"] == EDWARDS_ID)
].copy()

# Match records against Minnesota's games
reconciliation = min_games.merge(
    ant_records[
        [
            "game_id", "athlete_display_name", "team_abbreviation",
            "minutes", "points", "did_not_play", "reason"
        ]
    ],
    on="game_id",
    how="left",
    indicator=True
)

reconciliation["STATUS"] = np.select(
    [
        reconciliation["_merge"].eq("both") &
        reconciliation["minutes"].gt(0),

        reconciliation["_merge"].eq("both") &
        reconciliation["did_not_play"].eq(True),

        reconciliation["_merge"].eq("both")
    ],
    ["PLAYED", "DNP_RECORDED", "RECORD_NEEDS_REVIEW"],
    default="NO_EDWARDS_RECORD"
)

print("Minnesota unique games:", min_games["game_id"].nunique())
print("Edwards regular-season records:", len(ant_records))
print("Edwards unique game IDs:", ant_records["game_id"].nunique())

print("\nReconciled status counts:")
print(reconciliation["STATUS"].value_counts(dropna=False))

print("\nGames requiring investigation:")
display(
    reconciliation[
        reconciliation["STATUS"] != "PLAYED"
    ][
        [
            "game_id", "game_date", "athlete_display_name",
            "team_abbreviation", "minutes", "did_not_play",
            "reason", "STATUS"
        ]
    ].sort_values("game_date")
)

print("\nEdwards records not matching Minnesota's game IDs:")
display(
    ant_records[
        ~ant_records["game_id"].isin(min_games["game_id"])
    ][
        [
            "game_id", "game_date", "team_abbreviation",
            "minutes", "points", "did_not_play"
        ]
    ].sort_values("game_date")
)
```

**Edwards absensce stats**

```python


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. Select games classified as Edwards absences
# ------------------------------------------------------------

absence_game_ids = (
    team_games.loc[
        team_games["EDWARDS_STATUS"] == "MISSED",
        "game_id"
    ]
    .drop_duplicates()
    .tolist()
)

absence_data = wolves[
    wolves["game_id"].isin(absence_game_ids)
].copy()

# Exclude Anthony Edwards
absence_data = absence_data[
    absence_data["athlete_id"] != EDWARDS_ID
].copy()

# ------------------------------------------------------------
# 2. Prepare statistical columns
# ------------------------------------------------------------

stat_columns = [
    "minutes", "points", "assists",
    "rebounds", "blocks", "steals"
]

for col in stat_columns:
    if col in absence_data.columns:
        absence_data[col] = pd.to_numeric(
            absence_data[col],
            errors="coerce"
        )

# Count actual player appearances only
appearances = absence_data[
    absence_data["minutes"].notna()
    & (absence_data["minutes"] > 0)
].copy()

# ------------------------------------------------------------
# 3. Calculate per-game averages
# ------------------------------------------------------------

player_summary = (
    appearances.groupby(
        ["athlete_id", "athlete_display_name"],
        as_index=False
    )
    .agg(
        Games=("game_id", "nunique"),
        MPG=("minutes", "mean"),
        PPG=("points", "mean"),
        APG=("assists", "mean"),
        RPG=("rebounds", "mean"),
        BPG=("blocks", "mean"),
        SPG=("steals", "mean")
    )
)

# Require a minimum of 3 appearances
MIN_GAMES = 3

eligible_players = player_summary[
    player_summary["Games"] >= MIN_GAMES
].copy()

print("Games classified as Edwards absences:", len(absence_game_ids))
print("Players with at least", MIN_GAMES, "appearances:",
      len(eligible_players))

# ------------------------------------------------------------
# 4. Create the dashboard — 3 columns × 2 rows
# ------------------------------------------------------------

metrics = [
    ("MPG", "MINUTES PER GAME", "Minutes per game"),
    ("PPG", "POINTS PER GAME", "Points per game"),
    ("APG", "ASSISTS PER GAME", "Assists per game"),
    ("RPG", "REBOUNDS PER GAME", "Rebounds per game"),
    ("BPG", "BLOCKS PER GAME", "Blocks per game"),
    ("SPG", "STEALS PER GAME", "Steals per game")
]

fig, axes = plt.subplots(
    nrows=2,
    ncols=3,
    figsize=(24, 13)
)

axes = axes.flatten()

for ax, (metric, title, xlabel) in zip(axes, metrics):

    # Handle unavailable statistics
    if (
        metric not in eligible_players.columns
        or eligible_players[metric].notna().sum() == 0
    ):
        ax.axis("off")
        ax.text(
            0.5, 0.5,
            f"{title}\nData unavailable",
            ha="center",
            va="center",
            transform=ax.transAxes,
            fontsize=14
        )
        continue

    # Rank the top 8 players for a cleaner dashboard
    ranked = (
        eligible_players[
            eligible_players[metric].notna()
        ]
        .nlargest(8, metric)
        .sort_values(metric, ascending=True)
    )

    bars = ax.barh(
        ranked["athlete_display_name"],
        ranked[metric]
    )

    # Display average values beside the bars
    ax.bar_label(
        bars,
        labels=[f"{value:.1f}" for value in ranked[metric]],
        padding=3,
        fontsize=9
    )

    ax.set_title(
        title,
        fontsize=14,
        fontweight="bold",
        pad=12
    )

    ax.set_xlabel(xlabel, fontsize=10)
    ax.set_ylabel("")
    ax.tick_params(axis="y", labelsize=9)

    ax.grid(
        axis="x",
        linestyle="--",
        alpha=0.3
    )

    ax.set_axisbelow(True)

    max_value = ranked[metric].max()

    ax.set_xlim(
        0,
        max_value * 1.20 if max_value > 0 else 1
    )

fig.suptitle(
    "MINNESOTA TIMBERWOLVES",
    fontsize=23,
    fontweight="bold",
    y=0.99
)

fig.text(
    0.5,
    0.955,
    "WHO STEPPED UP DURING ANTHONY EDWARDS' ABSENCES?",
    ha="center",
    fontsize=15
)

fig.text(
    0.5,
    0.925,
    (
        f"2025–26 Regular Season | "
        f"{len(absence_game_ids)} classified absence games | "
        f"Per-appearance averages | "
        f"Minimum {MIN_GAMES} appearances"
    ),
    ha="center",
    fontsize=11
)

fig.text(
    0.5,
    0.015,
    "Player averages exclude games in which the player did not appear. "
    "Absence-game classification should be verified before publication.",
    ha="center",
    fontsize=9,
    style="italic"
)

plt.tight_layout(
    rect=[0.025, 0.045, 0.98, 0.90],
    w_pad=2.5,
    h_pad=3.0
)

# Save one high-resolution image for the article
dashboard_filename = "timberwolves_edwards_absence_dashboard.png"

plt.savefig(
    dashboard_filename,
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()

print("Dashboard saved as:", dashboard_filename)

# ------------------------------------------------------------
# 5. Final player summary table — averages only
# ------------------------------------------------------------

summary_columns = [
    "athlete_display_name",
    "Games",
    "MPG",
    "PPG",
    "APG",
    "RPG",
    "BPG",
    "SPG"
]

player_summary_final = (
    eligible_players[summary_columns]
    .sort_values("MPG", ascending=False)
    .round(2)
    .reset_index(drop=True)
)

player_summary_final.index = (
    player_summary_final.index + 1
)

player_summary_final.index.name = "Rank"

print("\nPLAYER SUMMARY — AVERAGES DURING EDWARDS' ABSENCES")

display(player_summary_final)
```
